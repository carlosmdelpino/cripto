## 1) NTRU y polinomios

NTRU trabaja con polinomios en un anillo de convolución. La idea general es que la clave pública se obtiene a partir de polinomios de pequeño grado y que la operación de convolución permite realizar el cifrado y el descifrado sin depender de la factorización de enteros. Esto marca una diferencia importante respecto a RSA, ya que NTRU se apoya en estructuras algebraicas más bien que en la dificultad de factorizar enteros grandes.

## 2) Comparación con RSA y ECC

- RSA y ECC se basan en problemas clásicos (factorización y logaritmo discreto).
- NTRU se apoya en estructuras algebraicas de anillos y se considera una familia prometedora frente a ataques cuánticos.
- En la práctica, la transición requiere evaluar rendimiento, tamaños de clave, compatibilidad con certificados X.509 y la evolución de software y protocolos.

## 3) Por qué necesitamos firmas post-cuánticas

Un sistema no puede quedarse solo con un KEM seguro si también requiere verificar firmas digitales y autenticar documentos. Por eso la transición incluye no solo cifrado, sino también firma digital, autenticación y validación de certificados. Los estándares NIST incluyen esquemas como ML-DSA y SLH-DSA, que permiten reemplazar firmas clásicas resistentes a ataques cuánticos.

## 4) Diferencia entre KEM, firma y certificado híbrido

- Un KEM permite establecer una clave compartida secreta entre dos partes.
- Una firma post-cuántica permite autenticar entidades y documentos.
- Un certificado híbrido mezcla credenciales clásicas y post-cuánticas para facilitar la transición gradual.

## 5) ¿Qué servicios priorizar?

Los servicios de mayor sensibilidad suelen ser la autenticación de usuarios, las VPN, la firma de documentos y la validación de certificados. También son críticos los sistemas con datos de alto valor, base de datos académica o procesos de matrícula y evaluación. Priorizar estos servicios permite identificar puntos de riesgo antes de desplegar el cambio general.

## 6) Plan de transición

1. Inventariar algoritmos y certificados en uso.
2. Añadir soporte híbrido para KEMs y firmas.
3. Probar entornos con software heredado y compatibilidad con certificados X.509.
4. Desplegar de forma progresiva, validando interoperabilidad y seguridad antes del reemplazo completo.

## 7) Implementación de ejemplo

```python
from math import gcd

N = 15
a = 7

values = []
for x in range(1, 10):
    values.append((x, pow(a, x, N)))

period = 0
for i in range(1, 10):
    if pow(a, i, N) == 1:
        period = i
        break

half = pow(a, period // 2, N)
p = gcd(half - 1, N)
q = gcd(half + 1, N)

print('Valores:', values)
print('Periodo r:', period)
print('a^(r/2) mod N =', half)
print('Factores:', p, q)
print('Resultado:', p * q)
```
