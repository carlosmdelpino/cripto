## 1) Diferencia entre bit y qubit

Un bit clásico solo puede estar en 0 o 1. Un qubit puede estar en una superposición `alpha|0> + beta|1>`, con `|alpha|^2 + |beta|^2 = 1`.

Esto permite que un algoritmo cuántico explore múltiples posibilidades de forma distinta a la computación clásica. En la práctica, la superposición es la base de muchos algoritmos cuánticos, pero no significa que el qubit sea una mezcla aleatoria sin interpretar; más bien, ofrece amplitudes sobre las que una medida posterior puede producir resultados con determinadas probabilidades.

## 2) Shor y la amenaza sobre RSA

El algoritmo de Shor reduce la factorización de un entero grande al problema de encontrar el periodo de una función modular. Si se encuentra ese periodo, entonces se pueden obtener los factores del número.

Como RSA se apoya en la dificultad de factorizar `n = p*q`, la computación cuántica rompe esa base. Esta es la razón por la que los mecanismos asimétricos clásicos son la principal preocupación cuando se analiza la transición a la criptografía post-cuántica.

## 3) Grover y la reducción de la seguridad simétrica

Grover ofrece una aceleración cuadrática frente a búsquedas no estructuradas. En la práctica, una clave simétrica de `k` bits pasa a tener seguridad equivalente a `k/2` bits.

Por eso, se recomienda usar tamaños mayores en AES o en derivados de clave cuando se considera la amenaza cuántica. No destruye el cifrado simétrico como Shor destruye RSA, pero sí reduce la seguridad efectiva.

## 4) Qué es ML-KEM

ML-KEM (Kyber) es un KEM post-cuántico basado en problemas de aprendizaje con errores. El objetivo es compartir una clave secreta que resista ataques cuánticos. En un esquema híbrido, la clave de sesión puede combinar un secreto clásico y un secreto post-cuántico para ofrecer compatibilidad y robustez durante la transición.

## 5) ¿Qué parte del sistema es crítica?

La parte más sensible se suele encontrar en las primitivas asimétricas: intercambio de claves, autenticación y firma de certificados. Si una entidad usa TLS clásico con algoritmo de clave pública vulnerable, toda la capa de sesión puede verse comprometida. Por esa razón, una estrategia realista prioriza la protección de la capa de intercambio de claves.

## 6) Conclusión

El principal punto del caso es que la amenaza cuántica no afecta por igual a todos los mecanismos. La criptografía asimétrica clásica es la más sensible; los algoritmos simétricos siguen siendo útiles, pero requieren tamaños mayores. Por ello, la migración a la PQC debe abordarse de forma estratégica, con protocolos híbridos y una planificación de criptoagilidad.

## 7) Implementación simple

```python
def mod_inverse(a, m):
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    return None

p = 5
q = 11
n = p * q
phi_n = (p - 1) * (q - 1)
e = 3
d = mod_inverse(e, phi_n)

message = 7
cipher = pow(message, e, n)
plain = pow(cipher, d, n)

print('d =', d)
print('cipher =', cipher)
print('plain =', plain)

for m in [1, 2, 7, 9]:
    c = pow(m, e, n)
    p = pow(c, d, n)
    print(f'm={m} -> c={c} -> p={p}')
```
