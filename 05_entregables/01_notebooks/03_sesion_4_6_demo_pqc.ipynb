# Demostración breve de criptografía post-cuántica

Este notebook reúne ejemplos sencillos para apoyar la explicación de los temas 4, 5 y 6.
Sirve para ilustrar LWE, la lógica de NTRU y la idea básica de firma con ML-DSA.

```python
# 1) Ejemplo simple de LWE
import numpy as np

q = 17
A = np.array([[4, 7], [3, 5]], dtype=int)
s = np.array([2, 6], dtype=int)
e = np.array([1, -1], dtype=int)
b = (A @ s + e) % q
print('A =
', A)
print('s =', s)
print('e =', e)
print('b =', b)
```

```python
# 2) Mini ejemplo de convolución circular para NTRU
N = 4
Q = 17

def cyclic_convolution(a, b, modulus):
    out = []
    for k in range(N):
        total = 0
        for i in range(N):
            total += a[i] * b[(k - i) % N]
        out.append(total % modulus)
    return out

r = [1, 0, 1, 0]
h = [9, 10, 1, 14]
c = cyclic_convolution(r, h, Q)
print('r*h =', c)
```

```python
# 3) Ilustración de la idea de firma ML-DSA
import numpy as np

A = np.array([[4, 7], [3, 5]], dtype=int)
s = np.array([2, 1], dtype=int)
t = (A @ s) % 17
y = np.array([1, 2], dtype=int)
w = (A @ y) % 17
c = 2
z = (y + c * s) % 17
verif = (A @ z - c * t) % 17
print('t =', t)
print('w =', w)
print('z =', z)
print('verif =', verif)
```

## Interpretación

- El ejemplo de LWE muestra que el atacante observa una ecuación con ruido, no un sistema exacto.
- En NTRU, la convolución circular es el mecanismo central para mezclar polinomios.
- En ML-DSA, la verificación compara si el reto reconstruido coincide con el hash del mensaje.
