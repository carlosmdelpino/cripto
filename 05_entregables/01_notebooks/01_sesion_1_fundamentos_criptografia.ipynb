## 1. Fundamentos

La criptografía permite proteger la información frente a accesos no autorizados. Sus objetivos básicos son:
- confidencialidad
- integridad
- autenticación
- no repudio

```python
# Ejemplo de cálculo con aritmética modular
modulo = 26
valor = 19
print(f'19 mod 26 = {valor % modulo}')
```

## 2. Función hash

Una función hash transforma una entrada de tamaño variable en una salida de longitud fija.
Ejemplo con SHA-256.

```python
import hashlib
mensaje = b'Hola, criptografia'
digest = hashlib.sha256(mensaje).hexdigest()
print('Hash SHA-256:', digest)
```

## 3. Ejercicio matemático sencillo

Calcula el módulo de los siguientes valores: 17, 29, 43 respecto a 10.

Puedes intentarlo con Python:

```python
for n in [17, 29, 43]:
    print(f'{n} mod 10 = {n % 10}')
```

## 4. Comparativa rápida

- Simétrica: misma clave
- Asimétrica: clave pública y privada
- Híbrida: combina ambas
